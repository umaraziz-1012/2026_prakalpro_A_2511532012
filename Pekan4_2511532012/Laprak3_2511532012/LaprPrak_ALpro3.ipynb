## 2.1 Dasar Teori
1. Input.
input adalah fungsi dalam python yang di gunakan untuk menerima data atau memasukkan data dari pengguna melalui keyboard
2. String.
String adalah tipe data yang digunakan untuk menyimpan teks atau kumpulan karakter.
3. Konstanta.
Konstanta adalah nilai yang digunakan dalam program dan umumnya dianggap tidak berubah selama program dijalankan.
4. Boolean.
bolean adalah tipe data yang menyatakan dua kondisi yaitu benar atau salah, tipe data ini sering di gunakan di percabangan atau perulangan program.
5. Numerik.
tipe data numerik digunakan untuk menyimpan nilai berupa angka. Dalam Python, tipe numerik yang umum digunakan adalah integer (int) untuk bilangan bulat, float untuk bilangan desimal, dan complex untuk bilangan kompleks


## 2.2 Kode Program

### 2.2.1 Program "Input_2511532012"


```python
lebar_2012 = input('Masukkan nilai Lebar : ')
panjang_2012 = input('Masukkan nilai Panjang : ')
luas_2012 = int(lebar_2012) * int(panjang_2012)
print("Luas Persegi Panjang adalah : ", luas_2012)
```

**Penjelasan :**
1. Bagian 1 dan 2 berguna untuk menginputkan data dimana nantik saat di jalankan lebar dan panjang bisa di inputkan nilainya
2. Bagian 3 berfungsi untuk melakukan operasi setelah input dimana nilai luas berasal dari lebar di kali dengan panjang
3. Bagian 4 berfungsi untuk menunjukkan output dari program

### 2.2.2 Program "String_2511532012"

### 2.2.3 Program "numerik1_2511532012"

```python
#create a variable with integer value
a_2012=100
print("the type of variable having value",a_2012,"is",type(a_2012))

#create a variable with float value
b_2012=10.2345
print("the type of variable having value",b_2012,"is",type(b_2012))

#create a variable with complex value
c_2012=100+3j
print("the type of variable having value",c_2012,"is",type(c_2012))
```

**Penjelasan :**
1. Bagian 1 membuat variabel dengan nilai 100 dimana ini bertipe int
2. Bagian 2 membuat variabel dengan nilai 10.2345 dimana ini bertipe float
3. Bagian 3 membuat varianle dengan nilai 100+3j dimana ini bertipe complex

### 2.2.4 Program "Boolean_2511532012"

```python
#deklarasi variabel
is_lulus_2012 = True
is_cumlaude_2012 = True

#Menggunakan bolean
nilai_2012 = 85
batas_lulus_2012 = 75

#Mnentukan nilai bolean dari kondisi
status_kelulusan = nilai_2012 >= batas_lulus_2012

print("=== Check Status Kelulusan ===")
print("Nilai:", nilai_2012)
print("apakah lulus?", status_kelulusan)
print("Status Kelulusan:", status_kelulusan)
if is_lulus_2012 and is_cumlaude_2012:
    print("Selamat! Anda lulus dengan predikat cumlaude.")
```

**Penjelasan :**
1. Bagian 1 dan 2 membuat variabel dengan nilai true yang menunjukkan nilai benar
2. Bagian 3 dan 4 membuat variabel nilai dan batas lulus yang menggunakan variable boolean
3. Bagian 5 membandingkan variable 3 dan 4 jika lebih besar maka status kelulusan bernilai true
4. Bagian 6 sampai 9 untuk menghasilkan output
5. bagian 10 merupakan boolean dimana jika dia lulus dan cumloude maka output yang di keluarkan adalah bagian 11

### 2.2.5 Program "Konstanta_2511532012"

```python
from typing import Final
PI: Final = 3.14
print("pi: %f" % PI)
jari_2012 = float(input('Masukkan nilai jari-jari : '))
luas_2012 = PI * jari_2012 * jari_2012
print("Luas lingkaran dengan jari-jari %.2f adalah %.2f" % (jari_2012, luas_2012))
```

**Penjelasan :**
1. Bagian 1 bertujuan untuk menandakan bahwa suatuu variable tidak bisa di ubah
2. Bagian 2 memasukkan nilai PI yang bersifat final 
3. Bagian 4 berguna untuk memasukkan nilai jari jari
4. Bagian 5 digunakan untuk menentukan nilai luas dimana nilainya berasal dari PI x jari jari x jari jari
5. Bagian 6 untuk menampilkan hasil bagian 5
