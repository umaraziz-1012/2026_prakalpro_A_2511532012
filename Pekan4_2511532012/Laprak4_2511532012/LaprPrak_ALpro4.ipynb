## 2.1 Dasar Teori
1. IF.
if merupakan percabangan yang digunakan untuk menjalankan suatu perintah apabila kondisi yang diberikan bernilai benar (True). Jika kondisi bernilai salah (False), perintah di dalam if tidak akan dijalankan. Percabangan if digunakan ketika program hanya membutuhkan pemeriksaan terhadap satu kondisi.
2. IF ELSE.
if else merupakan percabangan yang digunakan ketika terdapat dua kemungkinan kondisi. Jika kondisi pada if bernilai benar, program menjalankan perintah pada if. Sebaliknya, jika kondisi bernilai salah, program menjalankan perintah pada else.
3. IF ELIF ELSE.
if elif else digunakan ketika program memiliki lebih dari dua kemungkinan kondisi. if digunakan untuk memeriksa kondisi pertama, elif digunakan untuk memeriksa kondisi berikutnya, sedangkan else digunakan apabila seluruh kondisi sebelumnya bernilai salah.
4. Multi IF .
Multi if adalah penggunaan beberapa pernyataan if secara terpisah dalam satu program. Setiap if akan diperiksa secara independen sehingga lebih dari satu kondisi dapat terpenuhi dan dijalankan.
5. Case.
match-case merupakan struktur percabangan dalam Python yang digunakan untuk mencocokkan suatu nilai dengan beberapa kemungkinan pilihan. Struktur ini dapat digunakan sebagai alternatif ketika program memiliki banyak pilihan berdasarkan nilai tertentu.

### 2.2.1 Program "IF1_2511532012"


```python
ipk_2012 = float(input("input IPK Anda : "))
if ipk_2012 > 2.75:
    print ("ANDA LULUS SANGAT MEMUASKAN DENGAN IPK : ", str(ipk_2012))
print("Program selesai")
```

### 2.2.4 Program "Multi_IF2_2511532012"

```python
total_belanja_2012 = float(input("Masukkan total belanja (Rp): "))

input_member_2012 = input("Apakah Anda member? (y/n): ").strip().lower()
is_member_2012 = input_member_2012 in ['y', "ya"]

input_promo_2012 = input("Apakah kode promo valid? (y/n): ").strip().lower()
kode_promo_valid_2012 = input_promo_2012 in ["y", "ya"]

total_diskon_persen_2012 = 0

if total_belanja_2012 > 100000:
    total_diskon_persen_2012 += 10

if is_member_2012:
    total_diskon_persen_2012 += 5
    
if kode_promo_valid_2012:
    total_diskon_persen_2012 += 15
    
nominal_diskon_2012 = total_belanja_2012 * (total_diskon_persen_2012 / 100)
total_bayar_2012 = total_belanja_2012 - nominal_diskon_2012

print("\n--- [ RINCIAN PEMBAYARAN ] ---")
print(f"Total Diskon : {total_diskon_persen_2012}% (Rp{nominal_diskon_2012:,.0f})")
print(f"Total Bayar  : Rp{total_bayar_2012:,.0f}")

print(f"Total diskon yang anda dapatkan: Rp{total_diskon_persen_2012}%")
```

**Penjelasan :**
1. Bagian 1 sampai 5 untuk memasukkkan inputan atau nilai
2. Bagian 7 sampai 12 adalah untuk semua kondisii yang mungkin terjadi dan apa yang akan di lakukan jika sesuai dengan kondisi 
3. Bagian 13 dan 14 adalah untuk menentukan nilai dari nominal diskon dan total bayar bedasarkan kondisi yang sesuai
4. Bagian 15 sampai 18 adalah untuk ouputnya

### 2.2.5 Program "IF_ELIF_ELSE1_2511532012"

### 2.2.7 Program "Match_Case1_2511532012"

```python
bulan_2012 = int(input("Masukan angka bulan (1-12) : "))
match bulan_2012:
    case 1:
        print("Januari")
    case 2:
        print("Februari")
    case 3:
        print("Maret")
    case 4:
        print("April")
    case 5:
        print("Mei")
    case 6:
        print("Juni")
    case 7:
        print("Juli")
    case 8:
        print("Agustus")
    case 9:
        print("September")
    case 10:
        print("Oktober")
    case 11:
        print("November")
    case 12:
        print("Desember")
    case _:
        print("Angka Tidak Valid")
```

**Penjelasan :**
1. Bagian 1 untuk memmasukkan kita ingin case yang mana
2. Bagian kebawah adalah semua case yang ada, output sesuai dengan input yang di masukkan
