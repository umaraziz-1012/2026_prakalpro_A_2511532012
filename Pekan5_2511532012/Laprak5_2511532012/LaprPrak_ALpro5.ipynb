Perulangan merupakan salah satu konsep dasar yang penting dalam pemrograman. Perulangan digunakan untuk menjalankan suatu perintah secara berulang tanpa harus menuliskan perintah yang sama berkali-kali. Dengan adanya perulangan, program dapat dibuat lebih sederhana dan efisien.

## 2.1 Dasar Teori
1. Perulangan For
Perulangan for adalah perulangan yang digunakan untuk menjalankan suatu blok kode secara berulang berdasarkan suatu urutan atau rentang nilai. Perulangan ini berguna ketika jumlah perulangan sudah diketahui atau dapat ditentukan.

2. Fungsi Range()
range() digunakan untuk menghasilkan urutan bilangan yang dapat digunakan oleh perulangan for.
Bentuk umum:
range(start, stop, step)
start merupakan nilai awal, stop merupakan batas akhir yang tidak ikut dihitung, sedangkan step menentukan perubahan nilai setiap perulangan.

3. Nested For
Nesting atau nested for adalah penggunaan perulangan for di dalam perulangan for lainnya. Perulangan bagian dalam akan dijalankan setiap kali perulangan bagian luar berjalan.

4. Perulangan dengan Kondisi
Perulangan for dapat dikombinasikan dengan if untuk menentukan perintah yang dijalankan berdasarkan kondisi tertentu. Hal ini berguna dalam pembuatan pola yang memiliki bentuk berbeda pada bagian tertentu.


5. Perulangan untuk Membuat Pola
Perulangan dapat digunakan untuk mencetak karakter, angka, maupun spasi secara berulang. Dengan mengatur jumlah perulangan dan posisi karakter, dapat dibuat berbagai macam pola seperti segitiga, persegi, bingkai, dan pola jam pasir.

## 2.2 Kode Program

### 2.2.1 Program "perulangan_for1_2511532012"

```python
ulang_2012 = int(input("Massukkan jumlah perulangan : "))
for i_2012 in range(ulang_2012):
    print(f"Perulangan ke-{i_2012}")
```

**Penjelasan :**
1. ulang digunakan untuk menyimpan jumlah perulangan yang dimasukkan pengguna.
2. range(ulang) membuat urutan angka mulai dari 0 sampai satu angka sebelum nilai ulang_2012.
3. i digunakan sebagai variabel perulangan.
4. Setiap perulangan akan mencetak nomor perulangan.
5. Perulangan akan berhenti setelah mencapai jumlah yang ditentukan.

### 2.2.2 Program "perulangan_for2_2511532012"

```python
ulang_2012 = int(input("Massukkan jumlah perulangan : "))
print("perulangan ke-0 sampai ke-",ulang_2012-1)
for i_2012 in range(ulang_2012):
    print(i_2012, end=" ")
print()
print("Perulangan ke-1 sampai ke-", ulang_2012)
for i in range(1,ulang_2012+1):
    print(i, end=" ")
```

**Penjelasan :**
1. Program meminta jumlah perulangan dari pengguna.
2. Perulangan pertama menggunakan range(ulang) sehingga dimulai dari angka 0.
3. Perulangan kedua menggunakan range(1, ulang+1) sehingga dimulai dari angka 1.
4. end=" " digunakan agar angka ditampilkan pada baris yang sama dan dipisahkan dengan spasi.
5. Perbedaan kedua perulangan terdapat pada nilai awal dan batas akhirnya.

### 2.2.3 Program "perulangan_for3_2511532012"

```python
ulang_2012 = int(input("Massukkan jumlah perulangan : "))

jumlah_2012 = 0
for i_2012 in range(1, ulang_2012+1):
    print(i_2012, end=" ")
    jumlah_2012 = jumlah_2012 + i_2012

    if i_2012 < ulang_2012:
        print(" + ", end=" ")
    else:
        print(" = ", jumlah_2012,end=" ")
print()
print("jumlah", jumlah_2012)
```

**Penjelasan :**
1. ulang digunakan untuk menentukan batas perulangan.
2. jumlah diberi nilai awal 0 untuk menyimpan hasil penjumlahan.
3. Perulangan dimulai dari angka 1 sampai nilai ulang.
4. Setiap nilai i ditambahkan ke jumlah.
5. Kondisi if digunakan untuk menentukan apakah program mencetak tanda + atau tanda =.
6. Setelah semua perulangan selesai, jumlah keseluruhan ditampilkan.

### 2.2.4 Program "nested_for1_2511532012"

```python
batas_2012 =  int(input("Masukkan Nilai Batas : "))
for line_2012 in range(1, batas_2012 + 1):
    for j in range(1,(-1 * line_2012 + batas_2012)+1):
        print(".", end=" ")
    print(line_2012)
```

2. line merupakan perulangan bagian luar yang menentukan nomor baris.
3. j merupakan perulangan bagian dalam yang digunakan untuk mencetak titik.
4. Jumlah titik ditentukan oleh rumus:

### 2.2.5 Program "nested_for2_2511532012"

```python
batas_2012 =  int(input("Masukkan Nilai Batas : "))
for i_2012 in range(1, batas_2012 + 1):
    for j_2012 in range (1, batas_2012+1):
        print("*", end=" ")
    print()
```

**Penjelasan :**
1. batas digunakan untuk menentukan ukuran pola.
2. Perulangan luar i menentukan jumlah baris.
3. Perulangan dalam j menentukan jumlah karakter * pada setiap baris.
4. Karena jumlah perulangan bagian dalam selalu sama dengan batas, setiap baris memiliki jumlah * yang sama.
6. Hasil akhirnya membentuk pola persegi.

### 2.2.6 Program "nested_for3_2511532012"

```python
batas_2012 =  int(input("Masukkan Nilai Batas : "))
for i_2012 in range(batas_2012 + 1):
    for j_2012 in range(batas_2012+1):
        print(i_2012+j_2012, end=" ")
    print()
```

**Penjelasan :**
1. batas menentukan batas nilai perulangan.
2. i digunakan sebagai indeks baris.
3. j digunakan sebagai indeks kolom.
4. Nilai yang dicetak merupakan hasil dari: i + j
5. Setiap perubahan baris atau kolom akan menghasilkan nilai yang berbeda.
6. Nested for digunakan karena program mencetak data dalam bentuk baris dan kolom.

### 2.2.7 Program "nested_for4_2511532012"

```python
tinggi_2012 = int(input("Masukkan tinggi pola (bilangan genap, misal 10): "))

if tinggi_2012 % 2 != 0:
    print("Tinggi harus bilangan genap")
else:
    a_2012 = tinggi_2012
    c_2012 = a_2012
    lebar_2012 = (2 * tinggi_2012) - 2
    
    for i_2012 in range(1, tinggi_2012 + 1):
        b_2012 = c_2012 + 1
        
        for j_2012 in range(1, lebar_2012+1):
            
            if i_2012 == 1 or i_2012 == tinggi_2012:
                if j_2012 == 1 or j_2012 == lebar_2012:
                    print("#", end="")
                else:
                    print("=", end="")
            else: 
                if j_2012 == 1 or j_2012 == lebar_2012:
                    print("|", end="")
                else:
                    if j_2012 == c_2012:
                        print("<", end="")
                    elif j_2012 == b_2012:
                        print(">", end="")
                    elif j_2012 == (lebar_2012 - c_2012):
                        print("<", end="")
                    elif j_2012 == (lebar_2012 - c_2012 + 1):
                        print(">", end="")
                    elif j_2012 > b_2012 and j_2012 < (lebar_2012 - c_2012):
                        print(".", end="")
                    else:
                        print(" ", end="")
                        
        print()
            
        a_2012 -= 2
        
        if a_2012 <= 0:
            c_2012 = (-a_2012) + 2
        else:
            c_2012 = a_2012
```

**Penjelasan :**
1. Program meminta tinggi pola dari pengguna.
2. Kondisi tinggi % 2 != 0 digunakan untuk memastikan tinggi merupakan bilangan genap.
3. lebar menentukan lebar pola dengan rumus: lebar = (2 × tinggi) - 2
4. Perulangan i digunakan untuk membuat baris.
5. Perulangan j digunakan untuk membuat kolom.
6. Kondisi if digunakan untuk menentukan karakter yang dicetak pada setiap posisi.
7. "#" digunakan pada sudut bingkai.
8. = digunakan pada bagian atas dan bawah bingkai.
9. | digunakan sebagai sisi kiri dan kanan.
10. Karakter < dan > digunakan untuk membentuk bagian tengah pola.
11. Titik . digunakan untuk membentuk bagian dalam pola.
12. Nilai c diperbarui setiap selesai satu baris agar posisi pola dapat berubah.

### 2.2.8 Program "tugas_praktikum5_2511532012"

```python
tinggi_2012 = int(input("Masukkan tinggi segitiga: "))
for i_2012 in range(1, tinggi_2012 + 1):
    print(" " * (tinggi_2012 - i_2012), end="")
    for j_2012 in range(i_2012):
        if j_2012 == i_2012 - 1:
            print("*", end="")
        else:
            print("* ", end="")
    print()
```

**Penjelasan :**
1. tinggi digunakan untuk menentukan tinggi segitiga.
2. Perulangan luar i digunakan untuk menentukan jumlah baris.
3. Spasi di awal baris dihitung menggunakan rumus: tinggi - i
4. Perulangan dalam j digunakan untuk mencetak karakter *.
5. Kondisi if digunakan agar karakter * terakhir pada setiap baris tidak mendapatkan spasi tambahan.
6. Kombinasi spasi dan karakter * menghasilkan pola segitiga.

```python
ukuran_2012 = int(input("Masukkan Ukuran: "))

print("#" + "=" * (4 * ukuran_2012 + 5) + "#")

for baris_2012 in range(ukuran_2012, 0, -1):
    print("| ", end="")
    print(" " * (2 * (ukuran_2012 - baris_2012)), end="")

    for angka_2012 in range(baris_2012, 0, -1):
        print(angka_2012, end=" ")
    print("<*>", end="")

    for angka_2012 in range(1, baris_2012 + 1):
        print(" " + str(angka_2012), end="")
    print(" " * (2 * (ukuran_2012 - baris_2012)), end="")
    print(" |")

print("| " + " " * (2 * ukuran_2012 + 1) + "<*>" + " " * (2 * ukuran_2012 + 1) + " |")

for baris_2012 in range(1, ukuran_2012 + 1):
    print("| ", end="")
    print(" " * (2 * (ukuran_2012 - baris_2012)), end="")
    for angka_2012 in range(baris_2012, 0, -1):
        print(angka_2012, end=" ")
    print("<*>", end="")

    for angka_2012 in range(1, baris_2012 + 1):
        print(" " + str(angka_2012), end="")
    print(" " * (2 * (ukuran_2012 - baris_2012)), end="")
    print(" |")

print("#" + "=" * (4 * ukuran_2012 + 5) + "#")
```
